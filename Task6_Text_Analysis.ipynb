# Task 6: Text Analysis

This notebook analyses a short text. It counts the words, characters and vowels, and finds the words that repeat. These are basic text processing steps used in natural language processing (NLP).

```
# Sample text (manually created)
text = "Python makes data analysis simple. Data analysis helps people make better decisions. Python is simple to learn, and python is used everywhere."
print("Original text:")
print(text)
```

## Word and character counts

`split()` breaks the text into words on whitespace. `len()` counts the words and the characters.

```
words = text.split()
word_count = len(words)
character_count = len(text)
character_count_no_spaces = len(text.replace(" ", ""))

print("Words:", word_count)
print("Characters (including spaces):", character_count)
print("Characters (excluding spaces):", character_count_no_spaces)
```

## Vowel count

The text is converted to lowercase so that `A`, `E`, `I`, `O` and `U` are counted in the same way as their lowercase forms.

```
vowels = "aeiou"
vowel_count = sum(1 for ch in text.lower() if ch in vowels)
print("Vowels:", vowel_count)
```

## Repeated words

A `Counter` counts how often each word appears. Punctuation is removed and the text is lowercased first, so `Python` and `python` count as the same word.

```
import string
from collections import Counter

cleaned = text.lower().translate(str.maketrans("", "", string.punctuation))
word_counts = Counter(cleaned.split())
repeated = {word: count for word, count in word_counts.items() if count > 1}

print("Repeated words:")
for word, count in sorted(repeated.items(), key=lambda item: -item[1]):
    print(f"  {word}: {count}")
```

## Check against the original text

The counts are checked against the text directly, so the results are verified rather than assumed.

```
assert word_count == len(cleaned.split())
assert character_count == len(text)
assert vowel_count == sum(text.lower().count(v) for v in vowels)
assert repeated["python"] == 3 and repeated["simple"] == 2 and repeated["data"] == 2
print("All checks passed: the counts agree with the original text.")
```

## Answers to the interview questions

1. **What is text preprocessing?** Preparing raw text for analysis, usually by lowercasing it, removing punctuation, splitting it into words and removing noise such as extra spaces.
2. **How can repeated words be identified?** Split the text into words, normalise them (for example, lowercase and no punctuation), then count them with a dictionary or `collections.Counter`.
3. **What is NLP?** Natural language processing is the field of making computers understand and work with human language, such as text and speech. Counting words and finding patterns is one of its basic steps.
